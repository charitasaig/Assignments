```python
import pandas as pd
df=pd.read_csv("C:/Users/Pc/Downloads/employees.csv")
df1=pd.read_csv("C:/Users/Pc/Downloads/departments.csv")
print(df)
print(df1)
```

```python
#1
print(df.groupby("Department")["Salary"].mean())
```

```python
#2
print(df.groupby("Department")["EmployeeID"].count())
```

```python
#3
print(df.groupby("Department")["EmployeeID"].max())
```

```python
#6
print(df.groupby("Department")["Salary"].agg(["mean","min","max","sum"]))
```

```python
#7
print(df.groupby("Department")["Experience"].agg(["mean","min","max"]))
```

```python
#8
print(df.groupby("Department").agg({
    "Salary":["mean","max"],
    "Experience":["mean","min"]}))
```

```python
#9
print(df.groupby("Department").agg({
    "EmployeeID":["count"],
    "Salary":["sum"],
    "Experience":["max"]}))
    
```

```python
#10
inner=pd.merge(df,df1,on="Department",how="inner")
print(inner[["Name","Department","Manager","Budget"]])
```

```python
#11
inner=pd.merge(df,df1,on="Department",how="left")
#print(inner[["Name","Department","Budget"]])
print(inner[inner["Manager"].isnull()])
```

```python
#12
join=pd.merge(df,df1,on="Department",how="right")
print(join[join["Name"].isnull()])
```

```python
#13
join=pd.merge(df,df1,on="Department",how="outer")
print(join)
```

```python
#14
join=pd.merge(df,df1,on="Department",how="outer",indicator=True)
print(join)
print(join[join["_merge"]=="left_only"])
```

```python
#15
join=pd.merge(df,df1,on="Department",how="outer")
print("\naverage salary department-wise.\n")
print(join.groupby("Department")["Salary"].mean())
print("\ntotal employees in each department.\n")
print(join.groupby("Department")["EmployeeID"].count())
print("\nmanager of every department.\n")
print(join[["Department","Manager"]])
print("\ndepartment with the highest average salary.\n")
avg_salary=join.groupby("Department")["Salary"].mean()
print(avg_salary.idxmax())
print("\ndepartments that exist but have no employees.\n")
print(join[join["Name"].isnull()]["Department"])
print("\nDisplay the final merged DataFrame.\n")
print(join)
```
