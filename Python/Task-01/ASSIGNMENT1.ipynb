```python
'''
Problem 1 – Gaming Tournament
Display who secured
•
First Rank
•
Second Rank
•
Third Rank
without using the built-in sort() function.

Bonus Challenge
If two players have the same score, print
Tie Detected
instead of ranks.
'''

scores = [520, 890, 430]
players = ["Player1", "Player2", "Player3"]

if len(set(scores)) != len(scores):
    print("Tie Detected")
else:
    temp_scores = scores[:]
    temp_players = players[:]

    for i in range(len(temp_scores)):
        max_index = i
        for j in range(i + 1, len(temp_scores)):
            if temp_scores[j] > temp_scores[max_index]:
                max_index = j
        temp_scores[i], temp_scores[max_index] = temp_scores[max_index], temp_scores[i]
        temp_players[i], temp_players[max_index] = temp_players[max_index], temp_players[i]

    print("First Rank :", temp_players[0], temp_scores[0])
    print("Second Rank:", temp_players[1], temp_scores[1])
    print("Third Rank :", temp_players[2], temp_scores[2])
```

```python
'''Problem 3 – ATM Machine
Write a program that checks
• If enough balance exists, allow withdrawal.
• Otherwise print
Insufficient Balance
After successful withdrawal print the remaining balance.
Bonus
If the user tries to withdraw a negative amount,
display
Invalid Amount'''
balance = 15000
withdraw = 7000

if withdraw < 0:
    print("Invalid Amount")
elif withdraw <= balance:
    balance -= withdraw
    print("Remaining Balance:", balance)
else:
    print("Insufficient Balance")
```

```python
'''Problem 4 – Scholarship Selection
Students scoring
•
90 and above → Full Scholarship
•
75–89 → Half Scholarship
•
60–74 → Training Program
•
Below 60 → Better Luck Next Time

Display each student's scholarship category.'''
marks = [96,81,55,74,67,92]
for mark in marks:
    if mark >= 90:
        print(mark, "Full Scholarship")
    elif mark >= 75:
        print(mark, "Half Scholarship")
    elif mark >= 60:
        print(mark, "Training Program")
    else:
        print(mark, "Better Luck Next Time")
```

```python
''' Problem 6 – Shopping Cart
Find
•
Total Bill
•
Highest Price
•
Lowest Price
Do NOT use
sum()
max()
min()
Use loops and variables.'''
prices = [250,180,450,120,600]
total = 0
highest = prices[0]
lowest = prices[0]
for p in prices:
    total += p
    if p > highest:
        highest = p
    if p < lowest:
        lowest = p

print("Total:", total)
print("Highest:", highest)
print("Lowest:", lowest)
```

```python
''' Problem 17 – Cab Booking
Ride fare
₹850
Discount Rules
•
Above ₹1000 → 20%
•
Above ₹500 → 10%
•
Otherwise No Discount
Display final fare.'''
fare = 850

if fare > 1000:
    fare -= fare * 20 / 100
elif fare > 500:
    fare -= fare * 10 / 100

print("Final Fare:", fare)

```

```python
'''Problem 18 – School Attendance
Display
•
Total Present
•
Total Absent
using a loop.'''
attendance = ["Present","Absent","Present","Present","Absent"]

present = 0
absent = 0

for a in attendance:
    if a == "Present":
        present += 1
    else:
        absent += 1

print("Present:", present)
print("Absent:", absent)

```

```python
'''problem 20
1.
Check every system one by one.
2.
If any system reports "FAIL":
o
Display the system name.
o
Print "Launch Aborted!"
o
Stop checking further systems using break.
3.
If all systems are "OK", print:
Ready for Launch'''

systems = (
    "Engine",
    "Fuel",
    "Navigation",
    "Communication",
    "Cooling"
)

status = ["OK","OK","FAIL","OK","OK"]

for i in range(len(systems)):
    if status[i] == "FAIL":
        print(systems[i])
        print("Launch Aborted!")
        break
else:
    print("Ready for Launch")
```
