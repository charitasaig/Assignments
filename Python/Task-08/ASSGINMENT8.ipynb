```python
#1
import pandas as pd
marks=[30,48,29,58,49,56,33]
marks1=pd.Series(marks)
print("List:",marks)
print("\nSeries:\n")
print(marks1)
print("\nAdvantages of Series over list:\n")
print("1. Easy data analysis")
print("\n2. Built-in stastical functions")
print("\n3. Easy filtering and indexing")
```
